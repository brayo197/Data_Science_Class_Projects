# Cross validation In Regression

```python
#import libs
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 

#scikit libs
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.model_selection import train_test_split, KFold, cross_val_score

#Regression metrics
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error, root_mean_squared_error
```

```python
#load data
data = pd.read_csv("Advertising.csv")

#drop unamed column
del data["Unnamed: 0"]

#check top 5 rows
data.head()
```

```python
#check for null values
data.isna().sum().any()
```

```python
#check shape
data.shape
```

```python
#seperate label/target from features

#Features
X = data.drop("sales", axis=1)

#label
y = data["sales"]
```

```python
#split dataset into train and test set

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=41) #for reproducity

#confirm the shape
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
x_train[:4]
```

```python
#scale the features

#instantiate the std scaler
sc = StandardScaler()

#scale the xtrain
x_train_s = sc.fit_transform(x_train)

#tranform the x_test
x_test_s = sc.transform(x_test)

x_train_s[:4]
```

```python
# instatiate the model and train it

model = LinearRegression()

#train the model
model.fit(x_train_s,y_train)
```

```python
#check training score
round(model.score(x_train_s,y_train),2)
```

```python
#model prediction
y_pred = model.predict(x_test_s)

#check actual vs predicted
pd.DataFrame({"actual": y_test, "predicted": y_pred})[:5]
```

```python
#check model metricts
r2_score(y_pred, y_test)
```

```python
#MAE

mean_absolute_error(y_test, y_pred)
```

## crosss validation on Linear Regression

```python
#load data
data = pd.read_csv("Advertising.csv")

#drop unamed column
del data["Unnamed: 0"]

#check top 5 rows
data.head()
```

```python
#seperate label/target from features

#Features
X = data.drop("sales", axis=1)

sc= StandardScaler()
X_s = sc.fit_transform(X)

#label
y = data["sales"]
```

```python
#create a model
model2 =LinearRegression()
```

```python
#createing fold using kfold cv

kf = KFold(n_splits=5)
```

```python
#fit the model using cv with negative mse as metric or r2 metric
neg_mse_score = cross_val_score(model2, X_s, y, cv=kf, scoring='neg_mean_squared_error' )# or you cam use scoring of "r2'
neg_mse_score
```

```python
#MSE
mse = -neg_mse_score
mse
```

```python
#get one result(average mse)

np.mean(mse)
```

```python
#fit the model using cv with r2 metric
r2_score = cross_val_score(model2, X_s, y, cv=kf, scoring='r2' )# or you cam use scoring of "r2'
r2_score
```

```python
#get average r2 score
np.mean(r2_score)
```
