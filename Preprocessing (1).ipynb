```python
#import libraries
import pandas as pd 
import numpy as np 
import seaborn as sns

#import scikit
from sklearn.preprocessing import OneHotEncoder, LabelEncoder, \
                               StandardScaler, RobustScaler,MinMaxScaler
```

```python
#import the data

data = sns.load_dataset("tips")
data.head()
```

```python
#select cat coll

cat_coln = data.select_dtypes("category")
cat_coln
```

```python
#encode gender using ONE

#instantiate ohe
ohe = OneHotEncoder(sparse_output=False, dtype="int", drop="first")

gender_en = ohe.fit_transform(data[["sex"]])

gedner_en1 = pd.DataFrame(gender_en, columns=["sex_male"])
gedner_en1
```

```python
merged_df = pd.concat([data,gedner_en1],axis=1)
merged_df
```

```python
#drop gender column
del merged_df["sex"]

merged_df
```

```python
#option 2, using get dummies
day_dummy = pd.get_dummies(data["day"], dtype="int", drop_first=True)
day_dummy
```

```python
#one encoding multiple columns
data1 = sns.load_dataset("tips")
data1.head()
```

```python
#select cat colns
cat_colns = data1.select_dtypes("category")

cat_colns
```

```python
#OHE

cat_ohe = pd.get_dummies(cat_colns,dtype="int", drop_first=True)
cat_ohe
```

```python
#merged the data

merged_df1 = pd.concat([data1, cat_ohe], axis=1)

merged_df1
```

```python
#drop the cat colns

merged_df1.drop(cat_colns, axis=1, inplace=True)
merged_df1
```
